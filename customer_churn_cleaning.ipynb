```python
import pandas as pd

df = pd.read_csv("../Dataset/customer_churn.csv")

print("=" * 50)
print("CUSTOMER CHURN DATASET")
print("=" * 50)

print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Info:")
print(df.info())

print("\nRows and Columns:")
print(df.shape)
```

```python
print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\n========== DATA TYPES ==========")
print(df.dtypes)

print("\n========== SUMMARY STATISTICS ==========")
print(df.describe())

print("\n========== UNIQUE VALUES ==========")
print(df.nunique())
```

```python
df = df.drop_duplicates()

print("Rows after removing duplicates:", df.shape)

print(df.isnull().sum())

df = df.dropna()

print("Rows after removing missing values:", df.shape)

print(df.dtypes)

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

df = df.dropna()

df.to_csv("../Dataset/customer_churn_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
